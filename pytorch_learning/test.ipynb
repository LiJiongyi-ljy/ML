```python
import torch
import numpy as np
```

```python
x = torch.tensor([1.0,2.0,3.0])
y=x * 2
print(y)
```

```python
x = torch.tensor(3.0 , requires_grad=True)
y=x ** 2
y.backward()
print(x.grad)
```
