## Save and Load the Model

In this section we will look at how to persist model state with saving, loading and running model predictions.

````python
import torch
import torchvision.models as models
````

## Saving and Loading Model Weights

PyTorch models store the learned parameters in an internal state dictionary, called `state_dict`. These can be persisted via the `torch.save` method:

````python
model=models.vgg16(weights='IMAGENET1K_V1') #.vgg16 means create a VGG16 network(CNN) 
torch.save(model.state_dict(), 'model_weights.pth')  #torch.save(SAVED SOMETHING,PATH)
# ``.pth`` usually save the PyTorch data (often save weights, parameters, whole model object)
# ``.pt``
# ``.pkl`` Python pickle Serialization data
````

To load model weights, you need to create an instance of the same model first, and then load teh parameters using `load_state_dict()` method.

In the code below, we set `weights_only=True` to limit the functions executed during **unpickling**(反序列化) to only those necessary for loading weights. Using `weights_only=True` is considered a best practice when loading weights.

````python
model=models.vgg16() # we don't specify ``weight``. i.e. create untrained model
# Important : ``weights=None`` initializes the bias and weights instead of loading the pretrained weights of ImageNet
model.load_state_dict(torch.load('model_weights.pth', weights_only=True))
# Important : ``.load_state_dict`` loading the weights to the current model
# torch.load is the opposite of torch.save
model.eval()
````

Note

be sure to call `model.eval()` method before inferencing to set the dropout and **batch normalization** layers to evaluation mode. Failing to do this will yield inconsistent inference results.

## Saving and Loading Models with Shapes

When loading model weights, we needed to instantiate the model class first, because the class defines the structure of a network. We might want to save the structure of this class together with the model, in which case we can pass `model` (and not `model.state_dict()`) to the saving function:

````python
torch.save(model,'model.pth')
````

We can then load the models as demonstrated below.

As described in [Saving and loading torch.nn.Modules](https://pytorch.org/docs/main/notes/serialization.html#saving-and-loading-torch-nn-modules), saving `state_dict` is considered the best practice. However, below we use `weights_only=False` because this involves loading the model, which is a legacy use case for `torch.save`.

````python
model=torch.load('model.pth', weights_only=False)   #weights_only=False means the full pickle deserialization mechanism is allowed to be used.
# Under the hood, torch.load() uses PyTorch's serialization and deserialization mechanisms, which involve pickle.
# Meanwhile, weights_only=False means that the more complete Python object deserialization mechanism is permitted.
````

Note

This approach uses Python pickle module when serializing the model, thus it relies on the actual class definition to be available when loading the model.

pickle is a built-in Python module that can convert Python objects into byte data for storage, and can also restore the saved data back into Python objects.

```python
# assume we have a dictionary
student = {
    'name': 'Tom',
    'age': 18,
    'score': 95
}

# we can use pickle to store it on the disk
import pickle

# serialization: save the object
with open('student.pkl', 'wb') as f:
    pickle.dump(student, f)
```

Next time, if we want load the object

```python
import pickle

# deserizalization: load the object
with open('student.pkl', 'rb') as f:
    student = pickle.load(f)

print(student)
```
