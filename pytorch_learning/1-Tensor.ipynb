```python
import torch
import numpy as np
```

## Initialize a Tensor

```python
# Directly From data
data = [[1,2],[3,4]]
x_data = torch.tensor(data)
```

```python
#From a Numpy array
np_array=np.array(data)
x_data_from_np=torch.from_numpy(np_array)
```

```python
#From another tensor
x_ones=torch.ones_like(x_data)  #The new tensor retains the properties (shape, datatype) of the argument tensor, unless explicitly overridden.
print(x_data)
x_rand=torch.rand_like(x_data,dtype=torch.float)    #override the datatype of x_data ,and generate random values
print({x_rand})
```

```python
#With random or constant values
shape=(2,3) #shape is a tuple of tensor dimensions. In the functions below, it determines the dimensionality of the output tensor.
rand_tensor=torch.rand(shape)
ones_tensor=torch.ones(shape)
zeros_tensor=torch.zeros(shape)

print(rand_tensor)
print(ones_tensor)
print(zeros_tensor)
```

## Attributes of a Tensor

```python
#Tensor attributes describe their shape, datatype, and the device on which they are stored.
tensor=torch.rand(3,4)

print(tensor.shape)
print(tensor.dtype)
print(f"device tensor is stored on :{tensor.device}\n")
```

## Operations on Tensors

```python
#We need to explicitly move tensors to the accelerator using .to method (after checking for accelerator availability)
if torch.accelerator.is_available():
    tensor=tensor.to(torch.accelerator.current_accelerator())
```

```python
#indexing and slicing
tensor = torch.ones(4,4)

print(f"first row : {tensor[0]}")
print(f"first column : {tensor[:,0]}")
print(f"last column : {tensor[...,-1]}")    #numpy : arr[:,-1]
tensor[:,1] = 0
print(tensor)
```

```python
#Using torch.cat to join tensors
t1=torch.cat([tensor,tensor,tensor],dim=1)
print(t1)
```

```python
#Arithmetic operations
y1=tensor @ tensor.T    #@ computes the matirx multiplication
y2=tensor.matmul(tensor.T)  #matmul is the abbreviation of "Matrix Multiplication"

y3=torch.rand_like(y1)
torch.matmul(tensor,tensor.T,out=y3)

z1=tensor * tensor #* computes the element-wise product
z2=tensor.mul(tensor)

z3=torch.rand_like(tensor)
torch.mul(tensor,tensor,out=z3)
```

```python
#Single-element tensors
agg=tensor.sum()    #"agg" is a one-element tensor (the aggregation of tensor)
agg_item=agg.item()     #convert it to a Python numerical value using "item()"
print(agg_item,type(agg_item))
```

```python
#In-place operations
print(f"{tensor}\n")
tensor.add_(5)  #Operations that store the result into the operand are called in-place. They are denoted by a _ suffix. 
print(tensor)   #For example: x.copy_(y), x.t_() t_() means transfer
```

## Bridge with Numpy

Tensors on the CPU and NumPy arrays can share their underlying memory locations, and changing one will change the other.

### Tensor to NumPy array

```python
t=torch.ones(5)
print(f"t:{t}")
n=t.numpy()     #the numpy of t.numpy is the function of torch , not the np 
print(f"n:{n}")
```

```python
#A change in the tensor reflects in the NumPy array.
t.add_(1)
print(f"t:{t}")
print(f"n:{n}")
```

### NumPy array to Tensor

```python
n=np.ones(5)
t=torch.from_numpy(n)
```

```python
#Changes in the NumPy array reflects in the tensor
np.add(n,1,out=n)
print(f"t: {t}")
print(f"n: {n}")
```
